```python
import pandas as pd

students = pd.DataFrame({
    "name":  ["Ana", "Bruno", "Clara", "Diego", "Eva"],
    "grade": [85, 92, None, 78, 95],
    "hours_studied": [5, 8, 6, None, 9]
})

# 1. How many rows and columns does this DataFrame have?
# 2. How many missing values are in each column?
# 3. Fill the missing values with the column average
# 4. Filter and show only students with grade > 85

#1
students.shape
#2
students.isnull().sum()
#3
students.fillna(students.mean(numeric_only=True))
#4
students[students["grade"] > 85]


```

```python
import seaborn as sns

students = pd.DataFrame({
    "name":  ["Ana", "Bruno", "Clara", "Diego", "Eva"],
    "grade": [85, 92, 79, 78, 95],
    "hours_studied": [5, 8, 6, 4, 9]
})

sns.histplot(students["grade"], bins =5)
plt.show

sns.scatterplot(data = students, x = "hours_studied", y = "grade")
plt.show()

sns.heatmap(students[["grade","hours_studied"]].corr(), annot=True)
plt.show()


time = np.linspace(0,1,1000)
frequency = 400
wave = np.sin(2*np.pi * frequency * time)

plt.plot(time, wave)



scores = pd.DataFrame({
    "student": ["Ana", "Bruno", "Clara", "Diego", "Eva"],
    "math":    [88, 92, 75, 65, 95],
    "english": [70, 85, 80, 60, 90],
    "hours_studied": [6, 8, 7, 4, 9]
})

#1
x = np.array(scores["student"])
y = np.array(scores["math"])
plt.bar(x,y)

#2 

sns.scatterplot(data = scores, x = "hours_studied", y = "math")
plt.show()

#3
sns.heatmap(scores[["math","english","hours_studied"]].corr(), annot=True)
plt.show()

#4
time = np.linspace(0,1,1000)
frequency = 261
wave  = np.sin( 2*np.pi*frequency*time)

plt.plot(time,wave)

```
